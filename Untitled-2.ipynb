```python
import cv2
import matplotlib.pyplot as plt
import os

# 1. قراءة الصورة
image_path = r"C:\Users\PC-LAB1\Pictures\Screenshots\Screenshot 2026-07-29 124604.png"
image_bgr = cv2.imread(image_path)

if image_bgr is None:
    print("خطأ: تعذر قراءة الصورة!")
else:
    # تحويل الألوان لـ RGB للعرض بـ Matplotlib
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    
    # أبعاد ومركز الصورة
    height, width = image_rgb.shape[:2]
    center = (width // 2, height // 2)

    # قائمة الزوايا المراد اختبارها
    angles = [45, 90, 180, 270]
    rotated_images = {}

    os.makedirs("output", exist_ok=True)

    # تدوير الصورة لكل زاوية
    for angle in angles:
        matrix = cv2.getRotationMatrix2D(center, angle, 1.0)
        rotated = cv2.warpAffine(image_rgb, matrix, (width, height))
        rotated_images[angle] = rotated
        
        # حفظ كل نسخة
        cv2.imwrite(f"output/rotated_{angle}.jpg", cv2.cvtColor(rotated, cv2.COLOR_RGB2BGR))

    # عرض جميع النتائج في لوحة واحدة
    fig, axes = plt.subplots(1, 5, figsize=(18, 5))

    # الصورة الأصلية
    axes[0].imshow(image_rgb)
    axes[0].set_title("Original")
    axes[0].axis("off")

    # عرض الزوايا
    for i, angle in enumerate(angles, start=1):
        axes[i].imshow(rotated_images[angle])
        axes[i].set_title(f"Rotated {angle}°")
        axes[i].axis("off")

    plt.tight_layout()
    plt.show()
```

```python
import cv2
import numpy as np
import os

def add_salt_and_pepper_noise(image, salt_prob=0.02, pepper_prob=0.02):
    """دالة لإضافة ضوضاء الملح والفلفل عشوائياً على الصورة"""
    noisy_image = image.copy()
    
    # 1. إضافة الملح (نقاط بيضاء = 255)
    num_salt = np.ceil(salt_prob * image.size)
    coords = [np.random.randint(0, i - 1, int(num_salt)) for i in image.shape[:2]]
    noisy_image[tuple(coords)] = 255

    # 2. إضافة الفلفل (نقاط سوداء = 0)
    num_pepper = np.ceil(pepper_prob * image.size)
    coords = [np.random.randint(0, i - 1, int(num_pepper)) for i in image.shape[:2]]
    noisy_image[tuple(coords)] = 0
    
    return noisy_image

# 1. قراءة الصورة الأصلية
image_path = r"C:\Users\PC-LAB1\Desktop\immge processing\paper_noisy_camera_man.webp"
image = cv2.imread(image_path)

if image is None:
    print("خطأ: تعذر قراءة الصورة! تأكد من صحة المسار.")
else:
    # 2. إضافة ضوضاء الملح والفلفل (بنسبة 2% لكل منهما)
    noisy_img = add_salt_and_pepper_noise(image, salt_prob=0.02, pepper_prob=0.02)

    # 3. استخدام Median Filter لإزالة الضوضاء (حجم النواة 5x5)
    cleaned_img = cv2.medianBlur(noisy_img, 9)

    # 4. حفظ النتائج في مجلد output
    os.makedirs("output", exist_ok=True)
    cv2.imwrite("output/noisy_image.jpg", noisy_img)
    cv2.imwrite("output/cleaned_image.jpg", cleaned_img)

    # 5. عرض النتائج في نوافذ مستقلة بحجم كبير وقابل للتحكم
    windows = {
        "1. Original Image": image,
        "2. Noisy Image (Salt & Pepper)": noisy_img,
        "3. Cleaned Image (Median Filter Result)": cleaned_img
    }

    for win_name, img_data in windows.items():
        cv2.namedWindow(win_name, cv2.WINDOW_NORMAL)
        cv2.resizeWindow(win_name, 1024, 768)
        cv2.imshow(win_name, img_data)

    print("تمت العملية بنجاح! اضغط أي زر لإغلاق النوافذ...")
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import os

# 1. قراءة الصورة
image_path = r"C:\Users\PC-LAB1\Desktop\immge processing\paper_noisy_camera_man.webp"
image = cv2.imread(image_path)

if image is None:
    print("خطأ: تعذر قراءة الصورة! تأكد من صحة المسار.")
else:
    # --- الطريقة الأولى: مصفوفة لابلاس الموزونة (Sharpening Kernel) ---
    # مصفوفة احترافية لزيادة الحدة مع الحفاظ على توازن الإضاءة
    sharpen_kernel = np.array([
        [0, -1,  0],
        [-1, 5, -1],
        [0, -1,  0]
    ])
    laplacian_sharpened = cv2.filter2D(image, -1, sharpen_kernel)

    # --- الطريقة الثانية: Unsharp Masking (الطريقة الأفضل عالمياً لتحسين الوضوح) ---
    # 1. تطبيق تنعيم جاوص خفيف
    gaussian = cv2.GaussianBlur(image, (0, 0), 3)
    # 2. إبراز التفاصيل عبر الدمج الموزون (Original * 1.5 - Gaussian * 0.5)
    unsharp_sharpened = cv2.addWeighted(image, 1.5, gaussian, -0.5, 0)

    # حفظ النتائج
    os.makedirs("output", exist_ok=True)
    cv2.imwrite("output/laplacian_sharpened.jpg", laplacian_sharpened)
    cv2.imwrite("output/unsharp_sharpened.jpg", unsharp_sharpened)

    # عرض النتائج في نوافذ كبيرة وقابلة للتكبير
    windows = {
        "1. Original Image": image,
        "2. Laplacian Kernel Sharpened": laplacian_sharpened,
        "3. Unsharp Masking (Pro Quality)": unsharp_sharpened
    }

    for win_name, img_data in windows.items():
        cv2.namedWindow(win_name, cv2.WINDOW_NORMAL)
        cv2.resizeWindow(win_name, 1024, 768)
        cv2.imshow(win_name, img_data)

    print("اضغط أي زر لإغلاق النوافذ...")
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import os

# 1. قراءة الصورة
image_path = r"C:\Users\PC-LAB1\Desktop\immge processing\image_with_shapes_2.jpeg"
image = cv2.imread(image_path)

if image is None:
    print("خطأ: تعذر قراءة الصورة! تأكد من صحة المسار.")
else:
    output_img = image.copy()
    
    # 2. معالجة الصورة وتحويلها لتحديد الأشكال
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    # عتبة ثنائية وتطبيق الإغلاق المورفولوجي لضمان كمال الأشكال
    _, thresh = cv2.threshold(blurred, 60, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
    morphed = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

    # 3. استخراج الحدود (Contours)
    contours, _ = cv2.findContours(morphed, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    print(f"تم العثور على {len(contours)} عنصر.\n" + "="*50)

    for i, cnt in enumerate(contours, start=1):
        area = cv2.contourArea(cnt)
        
        # تجاهل الضوضاء والعناصر الصغيرة جداً
        if area < 300:
            continue

        perimeter = cv2.arcLength(cnt, True)

        # تقريب أضلاع الشكل (Douglas-Peucker Algorithm)
        epsilon = 0.02 * perimeter
        approx = cv2.approxPolyDP(cnt, epsilon, True)
        num_sides = len(approx)

        # حساب مركز الشكل بطريقة آمنة مع تعديل المفاتيح m10 و m01
        M = cv2.moments(cnt)
        if M["m00"] != 0:
            cx = int(M["m10"] / M["m00"])
            cy = int(M["m01"] / M["m00"])
        else:
            cx, cy = approx[0][0][0], approx[0][0][1]

        shape_label = ""

        # --- التمييز بين الأشكال وحساب قياسات كل شكل على حدة ---

        # أ) المثلث (3 أضلاع)
        if num_sides == 3:
            shape_label = "Triangle"
            cv2.drawContours(output_img, [approx], -1, (0, 255, 0), 2)  # رسم بلون أخضر[cite: 1]
            
            # حساب أطوال الأضلاع الثلاثة
            side_lengths = []
            for j in range(3):
                pt1 = approx[j][0]
                pt2 = approx[(j + 1) % 3][0]
                length = np.linalg.norm(pt1 - pt2)
                side_lengths.append(round(length, 1))
            
            print(f"الشكل #{i}: مثلث (Triangle)")
            print(f"  - المساحة: {area:.1f} px²")
            print(f"  - أطوال الأضلاع الثلاثة: {side_lengths} px")

        # ب) المربع أو المستطيل (4 أضلاع) - مع التعامل مع الأشكال المائلة
        elif num_sides == 4:
            # استخدام المستطيل الأدنى المائل (Rotated Bounding Box)
            rect = cv2.minAreaRect(cnt)
            box = cv2.boxPoints(rect)
            box = np.intp(box)
            
            (x_r, y_r), (w, h), angle = rect
            aspect_ratio = min(w, h) / max(w, h) if max(w, h) != 0 else 0
            
            if aspect_ratio >= 0.90:
                shape_label = "Square"
            else:
                shape_label = "Rectangle"

            # رسم إطار مائل ينطبق على أضلاع المستطيل أو المربع تماماً
            cv2.drawContours(output_img, [box], 0, (255, 0, 0), 2)  # رسم بلون أزرق[cite: 1]
            
            print(f"الشكل #{i}: {shape_label}")
            print(f"  - المساحة: {area:.1f} px²")
            print(f"  - العرض (Width): {round(w, 1)} px")
            print(f"  - الارتفاع (Height): {round(h, 1)} px")

        # ج) الدائرة (أكثر من 5 أضلاع)
        elif num_sides > 5:
            shape_label = "Circle"
            (x, y), radius = cv2.minEnclosingCircle(cnt)
            center = (int(x), int(y))
            radius = int(radius)

            cv2.circle(output_img, center, radius, (0, 0, 255), 2)  # رسم بلون أحمر للدائرة فقط[cite: 1]
            
            print(f"الشكل #{i}: دائرة (Circle)")
            print(f"  - المساحة: {area:.1f} px²")
            print(f"  - نصف القطر (Radius): {radius} px")

        # د) أشكال مضلعة أخرى (خماسي، سداسي...)
        else:
            shape_label = f"Polygon ({num_sides} sides)"
            cv2.drawContours(output_img, [approx], -1, (255, 255, 0), 2)
            print(f"الشكل #{i}: مضلع ذو {num_sides} أضلاع")
            print(f"  - المساحة: {area:.1f} px²")

        print("-" * 40)

        # كتابة نوع الشكل والمساحة فوق المركز مباشرة بخط محدد ودقيق
        cv2.putText(output_img, shape_label, (cx - 35, cy - 5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.45, (0, 0, 0), 2)
        cv2.putText(output_img, f"A:{int(area)}", (cx - 35, cy + 15),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.4, (0, 0, 0), 1)

    # 4. حفظ وعرض النتيجة
    os.makedirs("output", exist_ok=True)
    cv2.imwrite("output/classified_shapes.jpg", output_img)

    cv2.namedWindow("Classified Shapes", cv2.WINDOW_NORMAL)
    cv2.resizeWindow("Classified Shapes", 1024, 768)
    cv2.imshow("Classified Shapes", output_img)

    print("تم تحديث الكود وتحديد جميع الأشكال المائلة بدقة! اضغط أي زر لإغلاق النافذة...")
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
