```python
import cv2
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# 1. قراءة الصورة
img = Image.open(r"C:\Users\PC-LAB1\Desktop\immge processing\ed44b0c4f03794879c13e936928a7d69.jpg")
image = np.array(img)

# 2. تحويل الصورة إلى التدرج الرمادي (أن أنظمة الموازنة القياسية تعمل على قناة واحدة)
if image.ndim == 3:
    gray_image = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
else:
    gray_image = image

# 3. تطبيق موازنة الهستوجرام (Histogram Equalization)
equalized_image = cv2.equalizeHist(gray_image)

# 4. إعداد لوحة رسم مقسمة (2x2) لتوضيح النتائج
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# --- الصورة الأصلية وهستوجرامها ---
axes[0, 0].imshow(gray_image, cmap="gray")
axes[0, 0].set_title("Original Grayscale Image")
axes[0, 0].axis("off")

axes[0, 1].hist(gray_image.ravel(), bins=256, range=[0, 256], color="black", alpha=0.75)
axes[0, 1].set_title("Original Histogram")
axes[0, 1].set_xlabel("Pixel Value (0-255)")
axes[0, 1].set_ylabel("Pixel Count")
axes[0, 1].set_xlim([0, 256])

# --- الصورة بعد الموازنة وهستوجرامها ---
axes[1, 0].imshow(equalized_image, cmap="gray")
axes[1, 0].set_title("Equalized Image")
axes[1, 0].axis("off")

axes[1, 1].hist(equalized_image.ravel(), bins=256, range=[0, 256], color="blue", alpha=0.75)
axes[1, 1].set_title("Equalized Histogram")
axes[1, 1].set_xlabel("Pixel Value (0-255)")
axes[1, 1].set_ylabel("Pixel Count")
axes[1, 1].set_xlim([0, 256])

plt.tight_layout()
plt.show()
```

```python
import cv2
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# 1. قراءة الصورة
img = Image.open(r"C:\Users\PC-LAB1\Desktop\immge processing\ed44b0c4f03794879c13e936928a7d69.jpg")
image = np.array(img)

# 2. تحويل الصورة إلى التدرج الرمادي (Grayscale)
if image.ndim == 3:
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
else:
    gray = image

# -------------------------------------------------------------
# 3. تطبيق خوارزميات تحسين التباين والإضاءة
# -------------------------------------------------------------

# أ) Global Histogram Equalization
equalized = cv2.equalizeHist(gray)

# ب) CLAHE (Contrast Limited Adaptive Histogram Equalization)
clahe_obj = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
clahe_img = clahe_obj.apply(gray)

# ج) Contrast Stretching (Min-Max Normalization)
stretched = cv2.normalize(gray, None, alpha=0, beta=255, norm_type=cv2.NORM_MINMAX)

# د) Linear Brightness and Contrast (Alpha & Beta)
# alpha=1.3 (زيادة التباين بنسبة 30%)، beta=30 (زيادة السطوع 30 درجة)
linear_adj = cv2.convertScaleAbs(gray, alpha=1.3, beta=30)

# هـ) Gamma Correction (gamma = 1.5 لتفتيح المناطق المظلمة)
gamma_val = 1.5
lut = np.array([((i / 255.0) ** (1.0 / gamma_val)) * 255 for i in range(256)]).astype("uint8")
gamma_img = cv2.LUT(gray, lut)

# و) Log Transformation
c = 255 / np.log(1 + np.max(gray))
log_img = c * (np.log(1 + gray.astype(np.float64)))
log_img = np.array(log_img, dtype=np.uint8)


# -------------------------------------------------------------
# 4. عرض جميع النتائج في شبكة مقسمة (2x4)
# -------------------------------------------------------------

fig, axes = plt.subplots(2, 4, figsize=(18, 9))

# قائمة النتائج مع العناوين
results = [
    (gray, "1. Original Grayscale"),
    (equalized, "2. Global Equalization"),
    (clahe_img, "3. CLAHE (Adaptive)"),
    (stretched, "4. Contrast Stretching"),
    (linear_adj, "5. Linear (Alpha/Beta)"),
    (gamma_img, f"6. Gamma Correction (\u03B3={gamma_val})"),
    (log_img, "7. Log Transformation")
]

# عرض الصور في أجزاء الشبكة
for i, (img_data, title) in enumerate(results):
    row, col = divmod(i, 4)
    axes[row, col].imshow(img_data, cmap="gray")
    axes[row, col].set_title(title, fontsize=11, fontweight="bold")
    axes[row, col].axis("off")

# إخفاء الخانة الأخيرة الثامنة (السطر الثاني، العمود الرابع) لأنها فارغة
axes[1, 3].axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# 1. قراءة الصورة
img = Image.open(r"\\192.168.1.10\shear all\lena5.jpg")
image_rgb = np.array(img)

# 2. تحويل الصورة إلى جميع الأنظمة (Color Spaces)
image_gray   = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2GRAY)
image_hsv    = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)
image_lab    = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2LAB)
image_ycrcb  = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2YCrCb)
image_hls    = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HLS)
image_edges  = cv2.Canny(image_gray, 100, 200)

# 3. إعداد شبكة عرض تضمن 8 صور (2 أسطر × 4 أعمدة)
fig, axes = plt.subplots(2, 4, figsize=(16, 8))

# القائمة تحتوي على (الصورة، العنوان، خريطة الألوان)
images_to_show = [
    (image_rgb, "1. Original RGB", None),
    (image_gray, "2. Grayscale", "gray"),
    (image_hsv, "3. HSV Space", None),
    (image_lab, "4. LAB Space", None),
    (image_ycrcb, "5. YCrCb Space", None),
    (image_hls, "6. HLS Space", None),
    (image_edges, "7. Canny Edges", "gray")
]

# عرض الصور داخل الشبكة
for i, (img_data, title, cmap) in enumerate(images_to_show):
    row, col = divmod(i, 4)
    if cmap:
        axes[row, col].imshow(img_data, cmap=cmap)
    else:
        axes[row, col].imshow(img_data)
    axes[row, col].set_title(title, fontsize=12)
    axes[row, col].axis("off")

# إخفاء مربع الرسم الأخير الفارغ (الخانة الثامنة)
axes[1, 3].axis("off")

# تعديل المسافات وعرض النتيجة
plt.tight_layout()
plt.show()
```

```python
import numpy as np
from PIL import Image

# 1. قراءة الصورة وتحويلها إلى مصفوفة
img = Image.open(r"\\192.168.1.10\shear all\lena5.jpg")
image = np.array(img)

# إذا كانت الصورة ملونة RGB، تحويلها إلى رمادية أحادية القناة أولاً
if image.ndim == 3:
    # حساب المتوسط للقنوات الثلاث لتحويلها لرمادي
    image = np.mean(image, axis=2).astype(np.uint8)

# 2. حساب عدد البيكسلات التي قيمتها تساوي 110
count_110 = np.sum(image == 110)

# 3. حساب إجمالي عدد البيكسلات في الصورة (الارتفاع × العرض)
total_pixels = image.size

# 4. حساب الاحتمال (النسبة المئوية)
probability = count_110 / total_pixels

print(f"عدد البيكسلات بقوة 110: {count_110}")
print(f"إجمالي عدد البيكسلات: {total_pixels}")
print(f"احتمال ظهور البيكسل 110: {probability:.6f}")
print(f"النسبة المئوية: {probability * 100:.4f}%")
```
